**Objective**: Compare Linear Regression vs Polynomial Regression and understand when to use each

#### Dataset Loading

```python
# import the libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# load the dataset

salaryData = pd.read_csv("task1_polynomial_data.csv")

# view the first 5 rows
salaryData.head()
```

```python
# check the dataset info

salaryData.info()
```

```python
# split the dataset into feature and target variable
X = salaryData['Experience_Years'].values.reshape(-1,1)
Y = salaryData['Salary'].values
```

#### Data Visualization

```python
# plot a scatterplot of the data

plt.scatter(X, Y)
plt.title("Experience Years VS Salary")
plt.xlabel("Experience Years")
plt.ylabel("Annual Salary (Thousand Dollars)")
plt.show()
```

#### Model Training

```python
# Linear Regression

from sklearn.linear_model import LinearRegression

# create the regression object
linearModel =  LinearRegression()

# fit the Linear regression to the data
linearModel.fit(X,Y)
```

```python
# Polynomial Regression with degree=2

from sklearn.preprocessing import PolynomialFeatures

# create the polynomial object
poly_reg = PolynomialFeatures(degree = 2)
# transform the data into a polynomial of degree 2
X_poly = poly_reg.fit_transform(X)
# create a linear regression object
lin_regression = LinearRegression()
# fit the polynomial feature to the model
lin_regression.fit(X_poly, Y)
```

```python
# Polynomial Regression with degree=3

from sklearn.preprocessing import PolynomialFeatures

# create the polynomial object
poly_reg1 = PolynomialFeatures(degree = 3)
# transform the data into a polynomial of degree 3
X_poly1 = poly_reg1.fit_transform(X)
# create a linear regression object
lin_regression1 = LinearRegression()
# fit the polynomial feature to the model
lin_regression1.fit(X_poly1, Y)
```

```python
# Polynomial Regression with degree=4

from sklearn.preprocessing import PolynomialFeatures

# create the polynomial object
poly_reg2 = PolynomialFeatures(degree = 4)
# transform the data into a polynomial of degree 2
X_poly2 = poly_reg2.fit_transform(X)
# create a linear regression object
lin_regression2 = LinearRegression()
# fit the polynomial feature to the model
lin_regression2.fit(X_poly2, Y)
```

#### Model Results Visualization

```python
# visualize the linear regression results

plt.scatter(X,Y)
plt.plot(X, linearModel.predict(X), color="red")
plt.title("Salary vs Experience (Linear Regression)")
plt.xlabel("Experience Years")
plt.ylabel("Annual Salary (Thousand Dollars)")
plt.show()
```

```python
# visualize the polynomial degree 2 regression results

plt.scatter(X,Y)
plt.plot(X, lin_regression.predict(X_poly), color="red")
plt.title("Salary vs Experience (Polynomial Regression degree 2)")
plt.xlabel("Experience Years")
plt.ylabel("Annual Salary (Thousand Dollars)")
plt.show()
```

```python
# visualize the polynomial degree 3 regression results

plt.scatter(X,Y)
plt.plot(X, lin_regression1.predict(X_poly1), color="red")
plt.title("Salary vs Experience (Polynomial Regression degree 3)")
plt.xlabel("Experience Years")
plt.ylabel("Annual Salary (Thousand Dollars)")
plt.show()
```

```python
# visualize the polynomial degree 2 regression results

plt.scatter(X,Y)
plt.plot(X, lin_regression2.predict(X_poly2), color="red")
plt.title("Salary vs Experience (Polynomial Regression degree 4)")
plt.xlabel("Experience Years")
plt.ylabel("Annual Salary (Thousand Dollars)")
plt.show()
```

#### Predictions

```python
# Predicting using the Linear regression model


print(f"The salary prediction for someone with 8.5 years of experence is: ${linearModel.predict([[8.5]])[0]:.2f} thousand")
```

```python
# Predicting using the polynomial regression with degree 2 model


print(f"The salary prediction for someone with 8.5 years of experence is: ${lin_regression.predict(poly_reg.fit_transform([[8.5]]))[0]:.2f} thousand")
```

```python
# Predicting using the polynomial regression with degree 3 model


print(f"The salary prediction for someone with 8.5 years of experence is: ${lin_regression1.predict(poly_reg1.fit_transform([[8.5]]))[0]:.2f} thousand")
```

```python
# Predicting using the polynomial regression with degree 4 model


print(f"The salary prediction for someone with 8.5 years of experence is: ${lin_regression2.predict(poly_reg2.fit_transform([[8.5]]))[0]:.2f} thousand")
```
