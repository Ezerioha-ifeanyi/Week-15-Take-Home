##### SVR with Proper Feature Scaling

```python
# split the dataset into feature and target variable

ice_x = icecreamData["Temperature"].values.reshape(-1,1)
ice_y = icecreamData["Ice_Cream_Sales"].values.reshape(-1,1)

from sklearn.preprocessing import StandardScaler
ice_sc_X = StandardScaler()
ice_sc_Y = StandardScaler()
ice_X = ice_sc_X.fit_transform(ice_x)
ice_Y = ice_sc_Y.fit_transform(ice_y)
```

```python
# Fitting SVR to the dataset
from sklearn.svm import SVR
ice_svr = SVR(kernel='rbf')
ice_svr.fit(ice_X, ice_Y.ravel())
```

```python
# Visualising the SVR results
plt.scatter(ice_x, ice_y, label='Actual Data')
plt.plot(ice_x, ice_sc_Y.inverse_transform(ice_svr.predict(ice_X).reshape(-1,1)), color='red', label='SVR Fitted With Scaling')
plt.title("Ice Cream Sales VS Temperature (SVR With Feature Scaling)")
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")
plt.legend()
plt.show()
```


### **Why Scaling Made the Difference**

**Visual Improvement**  
The impact is striking. In contrast to the flat, near-useless prediction line seen in previous step, the scaled SVR model (red line) now closely follows the curvature of the actual data points (blue).  

- It successfully captures the **non-linear relationship**: sales remain relatively flat at lower temperatures and then accelerate sharply as temperature rises.  
- Predicted values now span a realistic range (~$229 to ~$764), aligning well with the observed sales values in the dataset.

**Technical Reason Behind the Improvement**  
Applying `StandardScaler` transformed both the input feature (`Temperature`) and the target (`Sales`) to a comparable scale (centered around zero with roughly similar variance).  

- **Distance-based algorithms restored**: SVR relies heavily on distance calculations in feature space. Without scaling, the enormous magnitude of sales dollars (~hundreds) completely dominated the much smaller temperature values (°C), effectively masking temperature’s influence. Scaling equalized their contributions, allowing the kernel to properly detect the temperature–sales pattern.  
- **Correct interpretation preserved**: By applying `.inverse_transform()` to the scaled predictions, the final plot returns to the original units (°C and $), proving that the model learned meaningful relationships in scaled space and can reliably map them back to real-world values.

**Key Takeaway**  
This experiment clearly demonstrates that **feature scaling is not merely helpful for Support Vector Regression; it is essential**. Without it, SVR fails to model even obvious non-linear trends. With proper scaling, we now have a functional, well-behaved model ready for practical predictions.

### **Why Decision Trees Outperform Linear Regression for This Dataset**

The visualizations and comparisons from above steps clearly demonstrate why **Decision Tree Regression** often proves superior to Linear Regression when modeling datasets with non-linear or piecewise behavior; such as the study-hours-to-exam-score relationship here.

#### **Capturing Non-Linear “Steps” & Threshold Effects (Visual Proof)**
- **Linear Regression’s Limitation**: The fitted line (green) imposes a single, constant slope across the entire range. It assumes exam scores improve at a fixed rate per additional study hour — an unrealistic simplification for most real-world learning curves.
- **Decision Tree’s Strength**: The high-resolution plot shows the tree producing a **staircase-like** pattern. This structure naturally captures **threshold effects** and discrete jumps: small differences in study time (e.g., 10 vs. 11 hours) may have little impact, but crossing a critical boundary (e.g., 12 or 20 hours) can move the student into a noticeably higher performance bracket. The tree faithfully preserves these local, non-smooth patterns that the linear model flattens away.

#### **Local vs. Global Modeling**
- **Linear Regression - Global Fit**: It optimizes a single equation to minimize error across the full dataset. A few extreme points (outliers or leverage points) in one region can distort predictions everywhere else.
- **Decision Trees - Local Adaptation**: By recursively partitioning the feature space into isolated regions (leaves), the tree makes independent predictions within each segment. The expected score for a student studying 40 hours is unaffected by data points at 5 hours. This localized logic makes trees far more robust to heterogeneous patterns, clusters, or regime shifts within the data.

#### **Complete Invariance to Feature Scaling**
- In Task 2, Support Vector Regression collapsed without proper scaling because distance-based algorithms are highly sensitive to feature magnitudes.
- **Decision Trees require no scaling whatsoever**. They make decisions based solely on whether a value exceeds a threshold (e.g., “Hours > 18.5?”), not on absolute distances or magnitudes. This scale-invariance simplifies preprocessing and makes trees especially practical when working with raw, unnormalized data.

#### **Superior Interpretability (“White-Box” Nature)**
- Decision Trees are inherently transparent and business-friendly. The logic can be explained in plain language to non-technical stakeholders:  
  *“The model predicts a score of 78 because the student studied between 22 and 24 hours.”*
- Linear Regression, by contrast, returns abstract coefficients (e.g., +2.3 points per hour) that are mathematically precise but much harder for non-experts to translate into intuitive, actionable insights.

**Bottom Line**  
For datasets exhibiting non-linear, threshold-based, or piecewise relationships; like the step-like improvements in exam performance with additional study hours, **Decision Trees provide a more faithful, robust, and interpretable representation** than a single global linear fit. Their ability to adapt locally, ignore scaling issues, and deliver human-readable logic makes them the stronger choice here.

  **Why it fails:** A straight line cannot represent exponential salary growth. The extrapolation to mid-level (6.5) produces an absurdly high value far outside the realistic range (~$160k–$180k). This is a textbook case of severe underfitting on non-linear data.

#### Baseline: Multiple Linear Regression

```python
# import the necessary libraries
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import math
```

```python
# Build and Train the Linear Regression
energy_lin_reg = LinearRegression()
energy_lin_reg.fit(X_train, y_train)

# Make Predictions on Test Set
energy_y_pred_lin = energy_lin_reg.predict(X_test)
```

```python
# Calculate R², MAE and RMSE Metrics
energy_r2_lin = r2_score(y_test, energy_y_pred_lin)
energy_mae_lin = mean_absolute_error(y_test, energy_y_pred_lin)
energy_rmse_lin = math.sqrt(mean_squared_error(y_test, energy_y_pred_lin))

print("Baseline: Multiple Linear Regression Metrics")
print(f"R² Score: {energy_r2_lin:.3f}")
print(f"MAE:      {energy_mae_lin:.3f}")
print(f"RMSE:     {energy_rmse_lin:.3f}")
```

#### Model 2: Support Vector Regression

```python
# import the libraries
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

# Applying Standard Scaling to the Feature varibles
energy_sc_X = StandardScaler()
energy_X_train_scaled = energy_sc_X.fit_transform(X_train)
energy_X_test_scaled = energy_sc_X.transform(X_test)
```

```python
# Build and Train the SVR with RBF kernel
energy_svr = SVR(kernel='rbf')
energy_svr.fit(energy_X_train_scaled, y_train)
```

```python
# Make Prediction on the test set
energy_y_pred_svr = energy_svr.predict(energy_X_test_scaled)

# 4. Calculate Metrics
energy_r2_svr = r2_score(y_test, energy_y_pred_svr)
energy_mae_svr = mean_absolute_error(y_test, energy_y_pred_svr)
energy_rmse_svr = math.sqrt(mean_squared_error(y_test, energy_y_pred_svr))

print("    SVR (Scaled) Metrics    ")
print(f"R² Score: {energy_r2_svr:.4f}")
print(f"MAE:      {energy_mae_svr:.4f}")
print(f"RMSE:     {energy_rmse_svr:.4f}")

# 5. Compare with Baseline
print(f"\n    Improvement over Baseline Model")
print(f"R² Change: {energy_r2_svr - energy_r2_lin:.4f}")
```

#### Model 3: Decision Tree Regression

```python
# import the libraries

from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import math
```

```python
# Define depths to test
depths = [3, 5, 10, None]

print(" Decision Tree Performance According to Depth ")

# define key variables
best_depth = None
best_r2 = -float('inf')
best_dtree_model = None

for depth in depths:
    # Train the model
    energy_dt_reg = DecisionTreeRegressor(max_depth=depth, random_state=42)
    energy_dt_reg.fit(X_train, y_train)
    
    # Make Predictions on the Test set
    y_pred_dt = energy_dt_reg.predict(X_test)
    
    # Evaluate the model performance
    energy_r2_dt = r2_score(y_test, y_pred_dt)
    energy_mae_dt = mean_absolute_error(y_test, y_pred_dt)
    energy_rmse_dt = math.sqrt(mean_squared_error(y_test, y_pred_dt))
    
    print(f"\n[Maximum Depth: {depth}]")
    print(f"R² Score: {energy_r2_dt:.4f}")
    print(f"RMSE:     {energy_rmse_dt:.4f}")
    
    # Save the best model
    if energy_r2_dt > best_r2:
        best_r2 = energy_r2_dt
        best_depth = depth
        best_dtree_model = energy_dt_reg

print(f"\n---> Best Depth: {best_depth} (R²: {best_r2:.4f})")
```

## Model Selection and Limitations

### Which model performs best on the test set?
The **Decision Tree Regression** (with `max_depth=7`) emerges as the strongest performer on the test set.  
It achieves the highest explanatory power with an **R² of 0.826** and the lowest prediction error (**RMSE ≈ $2.85**).  
This significantly outperforms both Linear Regression and Polynomial Regression (degree 2), which recorded much weaker results (R² ≈ 0.04 and RMSE ≈ $6.70 for the polynomial model). The Decision Tree's ability to model sharp, non-linear price movements through data-driven splits explains its superior fit to the volatile, discontinuous nature of stock price data.

### Discussion of overfitting concerns
Overfitting remains a critical risk, especially with tree-based models like Decision Trees.  

- Without constraints, Decision Trees can grow very deep, memorizing noise in the training data rather than learning generalizable patterns; leading to excellent in-sample performance but poor out-of-sample results.  
- In this case, we mitigated this by limiting `max_depth=7`, which strikes a reasonable balance between model complexity and generalization (as evidenced by the strong test-set performance).  
- Polynomial Regression (degree 2), by contrast, showed clear signs of overfitting: its wildly oscillating curve in the comparison plot indicates it captured training noise rather than true underlying trends, resulting in dramatically higher test error.  
- Regular monitoring via cross-validation, pruning, or ensemble methods (e.g., Random Forest) would further reduce this risk in production settings.

### Limitations of using regression for stock price prediction
Regression models (even non-linear ones) face fundamental challenges when applied to stock prices:

- **Efficient Market Hypothesis & Random Walk Behavior**: Stock prices often approximate a random walk, making consistent outperformance difficult. Linear assumptions fail to capture this near-random, non-stationary behavior.  
- **Non-linearity & Regime Shifts**: Markets exhibit abrupt regime changes, volatility clustering, and non-linear dynamics that simple regression (linear or low-degree polynomial) cannot adequately model, often leading to underfitting or instability.  
- **Look-Ahead Bias & Data Leakage**: As seen here, relying on intraday `High_Price` and `Low_Price` introduces unrealistic future knowledge; in live trading, these are unavailable until after close.  
- **Sensitivity to Outliers & Noise**: Financial data is noisy and outlier-prone (e.g., flash crashes, earnings surprises), which can heavily distort regression fits.  
- **Lagging Nature**: Regression models are inherently backward-looking and struggle with sudden structural breaks or black-swan events.  
Overall, while useful for exploratory analysis or pairs trading, regression alone is rarely sufficient for robust, high-frequency stock prediction.

### Additional data or features that might improve predictions
To move beyond the current limitations, incorporating richer, more informative features could substantially enhance model performance:

- **Technical Indicators**: Moving Averages (SMA, EMA), RSI, MACD, Bollinger Bands, and volume-based metrics (e.g., OBV) to capture momentum, overbought/oversold conditions, and trend strength.  
- **Lagged Features**: Properly lagged versions of previous closes, returns, volatility (e.g., ATR), or high/low ranges to enable true forward-looking prediction without leakage.  
- **Sentiment & News Data**: Investor sentiment scores from news, social media (e.g., Twitter/X), or earnings transcripts to account for behavioral and event-driven movements.  
- **Macroeconomic & Fundamental Variables**: Interest rates, inflation, sector indices, earnings surprises, or volatility indices (VIX) for broader context.  
- **High-Frequency / Alternative Data**: Order book depth, trading volume patterns, or even satellite-derived supply-chain signals for short-term edges.  
Combining these with time-series-aware models (e.g., LSTM, GRU, or Transformer architectures) often yields better results than tree-based regression alone.

### Recommendation for actual trading — Yes or No? Why?
**No, I would not recommend deploying these models (as currently constructed) for live, real-money trading.**  

While the Decision Tree shows promising test-set accuracy, several red flags make it unsuitable for production trading without major enhancements:

- **Look-ahead bias** from using same-day `High_Price` and `Low_Price` renders the model unusable in real time; predictions must rely solely on past/available data.  
- Stock markets are noisy, non-stationary, and influenced by unpredictable events; even strong backtest results frequently fail to generalize (overfitting to historical regimes).  
- No model here accounts for transaction costs, slippage, market impact, or risk management; critical elements that can turn a theoretically profitable strategy unprofitable.  
- Regulatory, ethical, and risk considerations (e.g., avoiding misleading retail traders) further caution against direct use.

These models serve as excellent **educational/demonstration tools** and can highlight non-linear patterns effectively. For actual trading, they should only form a small part of a larger system; combined with rigorous walk-forward validation, ensemble methods, risk controls, and preferably more advanced architectures (e.g., LSTM or reinforcement learning) trained on properly lagged, diverse features.

```python
# Import the necessary libraries
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the Linear Regression Object and fit the training data
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Make Predictions using the test set
y_pred_train_lr = lr_model.predict(X_train)
y_pred_test_lr = lr_model.predict(X_test)

# Create a function to calculate the model metrics
def calculate_metrics(y_true, y_pred, set_name):
    r2 = r2_score(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    
    print(f"{set_name} Set Metrics")
    print(f"R² Score: {r2:.4f}")
    print(f"MAE:      ${mae:,.2f}")
    print(f"MSE:      {mse:,.2f}")
    print(f"RMSE:     ${rmse:,.2f}\n")
    
    return r2, mae, rmse

print("Baseline: Multiple Linear Regression")
r2_train_lr, mae_train_lr, rmse_train_lr = calculate_metrics(y_train, y_pred_train_lr, "Training")
r2_test_lr, mae_test_lr, rmse_test_lr = calculate_metrics(y_test, y_pred_test_lr, "Test")

# Create a DataFrame to Compare first 5 rows of Predictions vs Actuals
comparison_df = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred_test_lr})
print("First 5 Prediction Sample")
print(comparison_df.head())
```

#### 2.2 Model 2: Polynomial Regression
- Create polynomial features (test degrees: 2, 3)
- For each degree:
  - Transform features
  - Train model
  - Calculate all metrics
  - Check for overfitting (compare train vs test R²)
- Select best polynomial degree
- **Note**: Be careful with high degrees - may cause overfitting

```python
# Import the necessary libraries for Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures

# Define degrees to test
degrees = [2, 3]

best_poly_degree = 0
best_poly_r2 = -float('inf')

print("Model 2: Polynomial Regression Results")

# create a loop to train and evaluate polynomial regression models for each degree
for d in degrees:
    print(f"\n Testing Polynomial Degree {d}")
    
    # Create the Polynomial Features object and transform the training and test data
    poly = PolynomialFeatures(degree=d)
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)
    
    # Train the Model
    poly_model = LinearRegression()
    poly_model.fit(X_train_poly, y_train)
    
    # Make Predictions on the Training and Test Set
    y_pred_train_poly = poly_model.predict(X_train_poly)
    y_pred_test_poly = poly_model.predict(X_test_poly)
    
    # Calculate the Model Metrics
    # (Re-using the helper function we defined in the previous cell)
    r2_train, _, _ = calculate_metrics(y_train, y_pred_train_poly, "Training")
    r2_test, _, _ = calculate_metrics(y_test, y_pred_test_poly, "Test")
    
    # Check for Overfitting
    print(f" Train/Test Gap: {r2_train - r2_test:.4f}")
    
    if r2_test > best_poly_r2:
        best_poly_r2 = r2_test
        best_poly_degree = d

print(f"\n Best Polynomial Degree: {best_poly_degree} (Test R²: {best_poly_r2:.4f})")
```

- **Polynomial Degree 2 collapses under overfitting**: It showed near-perfect training fit (**Train R² ~ 0.99**), but suffered a dramatic drop on the test set (**Test R² ~ 0.38**), producing a very large **train–test performance gap of 0.611**. The model was fitting imaginary curvature that does not exist in the true relationship.
- **Polynomial Degree 3 perfectly overfits the training data**: It reached an essentially perfect training score (**Train R² = 1.00**, training error = $0.00$), meaning it memorized the training points almost exactly (including noise). Yet its generalization remained inferior (**Test R² ~ 0.8566**) compared to the much simpler linear model.

### Why is there a Drop in Model Performance?

Several factors likely contributed to the noticeable decline in test-set performance after hyperparameter tuning:

- **Overly Conservative Regularization**  
  The grid search selected relatively restrictive settings (`max_depth = 5`, `min_samples_leaf = 8`). While these parameters effectively prevent overfitting, they appear to have made the model **too simple** in this case, resulting in mild underfitting. The tuned tree lost the ability to capture some of the subtler, yet meaningful patterns that the deeper (less regularized) default tree was able to model.

- **Small Dataset + Cross-Validation Instability**  
  With only 200 records, each 5-fold CV split contains roughly 32 samples in the validation fold. On such small folds, cross-validation scores can be noisy and unstable. This increases the chance that the grid search selects hyperparameters that look “safe” (low variance across folds) but are slightly suboptimal for the particular test set used for final evaluation.

- **Random Feature Subsampling**  
  The best configuration included `max_features = 'sqrt'`, meaning each split only considers a random subset of features (≈√number of features). In a small dataset where a few features (especially **Year**) are extremely dominant, randomly excluding a critical feature at an important split can significantly degrade predictive power; an effect that becomes more pronounced with aggressive subsampling.

**Bottom line:** The tuned model traded some predictive accuracy for better generalization robustness, but on this particular small dataset, the trade-off leaned too far toward simplicity, resulting in lower test performance than the untuned (deeper) baseline.
